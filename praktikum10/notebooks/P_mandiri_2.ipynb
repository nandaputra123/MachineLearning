```python
import pandas as pd
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score

# Data sesuai tabel soal
data = {
    "NIM": ["TI001","TI002","TI003","TI004","TI005","TI006","TI007","TI008","TI009","TI010"],
    "Actual": ["Lulus","Lulus","Lulus","Lulus","Lulus","Tidak Lulus","Tidak Lulus","Tidak Lulus","Tidak Lulus","Tidak Lulus"],
    "Predicted": ["Lulus","Lulus","Lulus","Tidak Lulus","Tidak Lulus","Lulus","Tidak Lulus","Tidak Lulus","Tidak Lulus","Tidak Lulus"]
}

df = pd.DataFrame(data)
df
```

```python
y_true = df["Actual"]
y_pred = df["Predicted"]

# Konversi label menjadi biner (Lulus = 1, Tidak Lulus = 0)
df_binary = df.replace({"Lulus": 1, "Tidak Lulus": 0}).infer_objects(copy=False)
y_true_bin = df_binary["Actual"]
y_pred_bin = df_binary["Predicted"]

cm = confusion_matrix(y_true_bin, y_pred_bin)
cm
```

```python
accuracy = accuracy_score(y_true_bin, y_pred_bin)
precision = precision_score(y_true_bin, y_pred_bin)
recall = recall_score(y_true_bin, y_pred_bin)

accuracy, precision, recall
```
